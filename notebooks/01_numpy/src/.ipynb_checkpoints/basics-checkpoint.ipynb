# NumPy Basics - Hello World

Learning array creation and basic operations.

```python
import numpy as np

# Hello World - Create array
arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Type:", type(arr))
print("Shape:", arr.shape)
print("Dtype:", arr.dtype)
```

## Array Operations

```python
# Basic math
print("Sum:", arr.sum())
print("Mean:", arr.mean())
print("Max:", arr.max())

# Element-wise operations
doubled = arr * 2
print("Doubled:", doubled)
```
